```python
ciphertext = "WKH TXLFN EURZQ IRA MXPSV RYHU WKH ODCB GRJ"

for shift in range(1, 26):
    result = ""

    for letter in ciphertext:
        if letter.isalpha():
            result += chr(
                (ord(letter) - ord('A') - shift) % 26 + ord('A')
            )
        else:
            result += letter

    print("Shift", shift, ":", result)
```
